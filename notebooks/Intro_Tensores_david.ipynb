### 1. Manipulación de datos con tensores

La mayoría de las estructuras modernas de deep learning van a usar una clase de tensores (se parece a NumPy pero con más cosas muy útiles para DL). Por ejemplo: MXNet, PyTorch y TensorFlow
   - diferenciación automática
   - aprovecha las GPUs para acelerar el cálculo numérico, mientras que librerias como NumPy usan solo el CPU. 

Estas dos propiedades hacen que los tensores sean perfectos para entrenar, codificar y ejecutar redes neuronales. 

CPU(Unidad Central de Procesamiento): procesador de propósito general, diseñado para manejar una amplia gama de tareas complejas y administrar el sistema operativo y las aplicaciones. 
GPU(Unidad de Procesamiento Gráfico): procesador especializado, optimizado para realizar cálculos paralelos masivos, especialmente aquellos relacionados con gráficos, visualizaciones y aprendizaje automático


¿Qué es un tensor?
representa un arreglo numérico (probablemente multidimensional). En una sola dimensión un tensor es un vector. Con dos ejes un tensor es una matrix y para $k>2$ solo le llamamos un tensor de orden k. 
El tensor va a ser la interfaz principal para almacenar y manipular datos en librerias de aprendizaje profundo.


```python
import torch 
# arange crea un vector espaciado uniformemente en el rango [0,n)
x = torch.arange(12, dtype=torch.float32)
# cada valor es un elemento del tensor, en este caso tenemos 12 elementos 
print(x)
print(x.numel())
print(x.shape)
# podemos cambiar la forma del tensor usando reshape
X = x.reshape(3,4)
print(X)
# podemos crear tensores de 0's
print(torch.zeros((2,3,4)))
# podemos crear tensores aleatorios con una Gaussiana (0,1)
print(torch.rand(3,4))
# podemos construir tensores usando listas por ejemplo
print(torch.tensor([[2, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]]))
```

##### indexing y slicing de tensores 
Como con una lista de python podemos usar una propiedad muy similar de los tensores

```python
print(X[-1], X[1:3])
# puedes reescribir en el tensor como en una lista en python, dando coordenadas
X[1,2] = 17
print(X)
```

##### Operaciones de tensores 

```python
## 1. operaciones escalares
print(torch.exp(x))
## 2. operaciones entre tensores del mismo tamaño 
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
print(x + y)
print(x - y)
print(x * y) 
print(x / y)
print(x ** y)
## 3. concatenacion de tensores: 
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
print(torch.cat((X, Y), dim=0))
print(torch.cat((X, Y), dim=1))
## 4. tensores logicos 
print(X == Y)
### 5. cantidad de elementos en un tensor 
print(X.sum())
```

##### Broadcasting 
Operaciones de tensores que no son exactamente del mismo tamaño: 
- expande los dos arreglos copiando los elementos sobre el eje de dimensión 1, de manera que después de la transformación los dos tensores son del mismo tamaño
- hace operaciones elemento por elemento en estos nuevos tensores

```python
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))
print(a,b)
# braodcasting va a producir un arreglo de 3x2  replicando la matriz a sobre las columnas
# y la matriz ve sobre las filas y sumandolas elemento por elemento 
# a expanded [[0,0], [1,1], [2,2]] y b expanded [[0,1],[0,1],[0,1]]
print(a + b)
```

##### Ahorrando memoria
Recordemos que en python la ejecución de operaciones puede provocar la asignación de memoria a los resultados. Si hacemos $Y = X + Y$ python desreferencia el tensor al que apuntaba Y y en su lugar lo apuntamos Y a la memoria recién asignada. Osea después de correr la suma ahora python apunta a Y en una nueva locación.
¿Por qué? porque python primero evalua $Y + X$ guardandolo en un lugar nuevo y después apunta al nuevo lugar en la memoria. 
¿Por qué no queremos que pase esto?
1. No queremos asignar memoria innecesariamente todo el tiempo, por ejemplo para los parámetros de los modelos de ML que se van a estar actualizando, no queremos nueva memoria cada que itera el SGD. 
2. Podríamos apuntar a los mismos parámetros desde múltiples variales. 
La buena noticia es que podemos solucionar esto haciendo operaciones in situ. 

```python
# Problema: 
before = id(Y)
Y = Y + X
id(Y) == before
# Solucion:  
Z = torch.zeros_like(Y)
print('id(Z):', id(Z))
Z[:] = X + Y
print('id(Z):', id(Z))
# también podemos usar  X += Y si no vamos a usar X después
before = id(X)
X += Y
id(X) == before
```

### 3. Diferenciación automática 

Todas las librerias para DL tienen diferenciación automática (autograd).  A medida que pasamos datos a través de cada función sucesiva, la libreria construye un grafo que rastrea cómo cada valor depende de los otros. Para calcular las derivadas, la diferenciación automática funciona a la inversa a través de este grafo aplicando la regla de la cadena. (Recordemos forward y backward passing de las redes neuronales superficiales). 

```python
### Vamos a difrenciar la funcion y = 2x'x
x = torch.arange(4.0)
print(x)

# le asignamos un lugar para guardar nuestra diferenciación
print(x.requires_grad_(True))
# por default el gradiente es None 
print(x.grad)

# nuestra funcion es: 
y = 2 * torch.dot(x, x)
print(y)
# este methodo va a calcular la parcial de y con respecto a x
y.backward()
print(x.grad)

## sabemos que el graidente tiene que ser 4x
print(x.grad == 4*x)


## detaching: ejemplo si tenemos z = x * y , y = x * x pero queremos calcular la influencia directa de x en z, podemos 
## crear una nueva variable que borre como fue creada y. Asi que no va a haber ancentro en el grafo, por ejemplo
## sacar el gradiente de z con respecto a x, seria y, y no 3x**2

print(x.grad.zero_())
y = x * x
u = y.detach()
z = u * x
z.sum().backward()
x.grad == u

## pero todavia podemos calcular el gradiente de y con respecto de x: 
x.grad.zero_()
y.sum().backward()
x.grad == 2 * x
```
